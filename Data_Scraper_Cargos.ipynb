```python
df["Phone_Number"] = pd.to_numeric(
    df["Phone_Number"],
    errors="coerce"
).astype("Int64")

print(df["Phone_Number"].dtype)
```

```python
df["Amenities"] = (
    df[["Amenities_1", "Amenities_2", "Amenities_3"]]
    .fillna("")
    .astype(str)
    .apply(
        lambda row: ",".join(
            value.strip() for value in row if value.strip()
        ),
        axis=1
    )
)
```

```python
df = df.dropna(subset=["Phone_Number"])
```

```python
# ==== Clean Location safely ====
def extract_location(location):
    if pd.isna(location):
        return None

    location = str(location).strip()
    return location if location else None

df["Location"] = df["Location"].apply(extract_location)

# ==== Clean and standardize ====
df["Location"] = (
    df["Location"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

# ==== Group by Location and count records ====
location_counts = (
    df.groupby("Location")
    .size()
    .reset_index(name="Total_Records")
)

# ==== Sort by total records (descending) ====
location_counts = location_counts.sort_values(
    by="Total_Records",
    ascending=False
)

# ==== Show result ====
print(location_counts)
```

```python
# ====== Enter any location ======
target_location = input("Enter location (e.g., Chennai, Coimbatore, Trichy): ").strip()

# ====== Filter data ======
filtered_df = df[
    df["Location"].astype("string").str.contains(
        target_location,
        case=False,
        na=False,
        regex=False
    )
]

# ====== Save filtered data to Excel ======
output_file = f"{target_location}_Cargos_Companies.xlsx"
filtered_df.to_excel(output_file, index=False)

# ====== Display results ======
print(f"File created: {output_file}")
print(f"Total records found: {len(filtered_df)}")

from google.colab import files
files.download(output_file)
```

```python
# Step 1: Get input from the user
location = input("Enter location: ").strip()
amenities = input("Enter amenities: ").strip()

# Step 2: Filter the dataset
filtered_df = df[
    df["Location"].str.contains(
        location, case=False, na=False, regex=False
    ) &
    df["Amenities"].str.contains(
        amenities, case=False, na=False, regex=False
    )
]

# Step 3: Display results
print("Total records found:", len(filtered_df))
display(filtered_df)

# Step 4: Save as Excel
output_file = f"{location}_{amenities}_Companies.xlsx"
filtered_df.to_excel(output_file, index=False)

# Step 5: Download the file
files.download(output_file)
```

```python
import pandas as pd
import re


def create_company_link(name):
    if pd.isna(name):
        return ""

    # Replace spaces and special characters with hyphens
    name = str(name).strip().upper()
    name = re.sub(r'[^A-Z0-9]+', '-', name)
    name = name.strip('-')

    return f"https://{name}/"

# Change Company_Name to your actual company name column
df["Company_Link"] = df["Company_Name"].apply(create_company_link)

display(df[["Company_Name", "Company_Link"]].head(10))
```
