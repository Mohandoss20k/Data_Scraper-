```python
import re
import pandas as pd

# ==== Clean Location safely ====
def extract_location(location):
    if pd.isna(location):
        return None

    location = str(location).strip()
    return location if location else None

df["location"] = df["location"].apply(extract_location)

# ==== Clean and standardize ====
df["location"] = (
    df["location"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

# ==== Group by Location and count records ====
location_counts = (
    df.groupby("location")
    .size()
    .reset_index(name="Total_Records")
)

# ==== Sort by total records (descending) ====
location_counts = location_counts.sort_values(
    by="Total_Records",
    ascending=False
)

# ==== Show result ====
print(location_counts)
```

```python
district = input("Enter district: ").strip()

filtered_df = df[
    df["location"].str.split(",").str[-1].str.strip().str.casefold()
    == district.casefold()
]

display(filtered_df)

print("Total records:", len(filtered_df))
```

```python
import pandas as pd

# Count the last part of each location
last_word_counts = (
    df["location"]
    .dropna()
    .str.rsplit(",", n=1)
    .str[-1]
    .str.strip()
    .str.title()
    .value_counts()
    .reset_index()
)

# Rename the output columns
last_word_counts.columns = ["Last_Location", "Count"]

# Display the result
display(last_word_counts)
```
