```python
from urllib.parse import urlparse
import pandas as pd

def extract_short_url(url):
    if pd.isna(url):
        return ""

    path = urlparse(str(url)).path.strip("/")
    parts = path.split("/")

    if len(parts) >= 2:
        return "/" + "/".join(parts[:2])

    return ""

df["Short_URL"] = df["Website"].apply(extract_short_url)
```

```python
import pandas as pd
from urllib.parse import urlparse

def clean_url(url):
    if pd.isna(url):
        return ""

    url = str(url).strip()

    # Extract the URL path
    parsed = urlparse(url)

    # Remove the ID and tracking parameters, keeping company name
    parts = parsed.path.strip("/").split("/")

    if len(parts) >= 2:
        return f"{parsed.scheme}://{parsed.netloc}/{parts[0]}/{parts[1]}/"

    return url

# Replace Website with your actual Excel column name
df["Website"] = df["Website"].apply(clean_url)

print(df["Website"].head(10))
```

```python
import pandas as pd
import re

def clean_url(url):
    if pd.isna(url):
        return ""

    url = str(url).strip()

    match = re.search(
        r'https?://[^/]+/([^/?]+/[^/?]+)',
        url
    )

    if match:
        return "https://www.justdial.com/" + match.group(1).rstrip("/") + "/"

    return url

# Change Website to your actual column name
df["Website"] = df["Website"].apply(clean_url)

# Display the cleaned links without downloading
display(df[["Website"]].head(20))
```

```python
import pandas as pd
import re

def create_company_link(name):
    if pd.isna(name):
        return ""

    # Replace spaces and special characters with hyphens
    name = str(name).strip().upper()
    name = re.sub(r'[^A-Z0-9]+', '-', name)
    name = name.strip('-')

    return f"https://{name}/"

# Change Company_Name to your actual company name column
df["Company_Link"] = df["Company_Name"].apply(create_company_link)

display(df[["Company_Name", "Company_Link"]].head(10))
```

```python
import re
import pandas as pd

# ==== Clean Location safely ====
def extract_location(location):
    if pd.isna(location):
        return None

    location = str(location).strip()
    return location if location else None

df["Location"] = df["Location"].apply(extract_location)

# ==== Clean and standardize ====
df["Location"] = (
    df["Location"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

# ==== Group by Location and count records ====
location_counts = (
    df.groupby("Location")
    .size()
    .reset_index(name="Total_Records")
)

# ==== Sort by total records (descending) ====
location_counts = location_counts.sort_values(
    by="Total_Records",
    ascending=False
)

# ==== Show result ====
print(location_counts)
```
